```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

#  Model load 
model = tf.keras.models.load_model("crop_disease_model.keras")

#  Apni actual class names 
class_names = ['Corn___Common_rust', 'Corn___healthy', 
               'Potato___Early_blight', 'Potato___healthy',
               'Tomato___Early_blight', 'Tomato___healthy']  
# Prediction function
def predict_disease(img_path):
    img_size = (160, 160)
    img = tf.keras.utils.load_img(img_path, target_size=img_size)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, 0)
    img_array = preprocess_input(img_array)
    
    prediction = model.predict(img_array, verbose=1)
    predicted_class = class_names[np.argmax(prediction)]
    confidence = np.max(prediction) * 100
    
    return {
        "disease": predicted_class,
        "confidence": round(float(confidence), 2)
    }

#  Test 
result = predict_disease(r"C:\Users\great\Downloads\istockphoto-2208708679-612x612.jpg")
print(result)
```

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv(override=True)

API_KEY = os.getenv("OPENWEATHER_API_KEY")

print("Loaded:", bool(API_KEY))
print("Length:", len(API_KEY) if API_KEY else 0)
print("First 4:", API_KEY[:4] if API_KEY else None)
print("Last 4:", API_KEY[-4:] if API_KEY else None)

response = requests.get(
    "https://api.openweathermap.org/data/2.5/weather",
    params={
        "lat": 28.5355,
        "lon": 77.3910,
        "appid": API_KEY,
        "units": "metric"
    }
)

print("\nStatus:", response.status_code)
print("Response:", response.json())
```

```python
def get_temperature(city):
    params = {
        "q": city,
        "appid": API_KEY,
        "units": "metric"
    }

    response = requests.get(
        "https://api.openweathermap.org/data/2.5/weather",
        params=params
    )

    response.raise_for_status()
    data = response.json()

    return {
        "location": data["name"],
        "temperature": data["main"]["temp"],
        "humidity": data["main"]["humidity"]
    }
```

```python
get_temperature("Sector 63, Noida")
```

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv(override=True)

API_KEY = os.getenv("DATA_GOV_API_KEY")


def get_market_prices(limit=10, offset=0):

    url = "https://api.data.gov.in/resource/35985678-0d79-46b4-9ed6-6f13308a1d24"

    params = {
        "api-key": API_KEY,
        "format": "json",
        "limit": limit,
        "offset": offset
    }

    response = requests.get(url, params=params)

    response.raise_for_status()

    return response.json()
```

```python
data = get_market_prices()

print(data)
```
