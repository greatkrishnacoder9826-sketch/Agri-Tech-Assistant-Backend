```python
import whisper
model_whisper = whisper.load_model("small")
def transcribe_audio(audio_path):
    result = model_whisper.transcribe(audio_path, language="hi")
    return result['text']

text = transcribe_audio(r"C:\Users\great\Downloads\Watch this before Eating Green Leafy Vegetables at Night!.mp3")
print(text)
```
